```python
from matplotlib import pyplot as plt
from sklearn.datasets import make_checkerboard

n_clusters = (4, 3)
data, rows, columns = make_checkerboard(
    shape=(48, 48), n_clusters=n_clusters, noise=10, shuffle=False, random_state=42
)

print(data)


plt.matshow(data, cmap=plt.cm.Blues)
plt.title("Original dataset")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures


def true_fun(X):
    return np.cos(1.5 * np.pi * X)


np.random.seed(0)

trial, n_samples = 5, 30
degrees = [1, 4, 15]
X = np.random.rand(trial, n_samples)
Y = true_fun(X) + np.random.randn(trial, n_samples) * 0.1
X_plot = np.linspace(0, 1, 100)

with plt.style.context('Solarize_Light2'):
    plt.figure(figsize=(20, 14))

    for t in range(trial):
        x, y = X[:, t], Y[:, t]
        ax = plt.subplot(len(degrees) + 1, trial, t + 1)
        plt.plot(X_plot, true_fun(X_plot), label="true func")
        plt.scatter(x, y, s=20, label="samples")
        plt.xlim((0, 1))
        plt.ylim((-1.5, 1.5))

    for i in range(len(degrees)):

        x, y = X[:, 0], Y[:, 0]
        ax = plt.subplot(len(degrees) + 1, trial, (i + 1) * trial + 1)
        polynomial_features = PolynomialFeatures(degree=degrees[i], include_bias=True)
        xx = polynomial_features.fit_transform(x[:, np.newaxis])
        linear_regression0 = LinearRegression()
        linear_regression0.fit(xx, y)
        plt.plot(X_plot, true_fun(X_plot), label="true func")
        plt.plot(X_plot, linear_regression0.predict(polynomial_features.fit_transform(X_plot[:, np.newaxis])), label="model")
        plt.scatter(x, y, s=20, label="samples")
        plt.xlim((0, 1))
        plt.ylim((-1.5, 1.5))
        plt.legend(loc="best", labelcolor='#073642')

        x, y = X[:, 1], Y[:, 1]
        ax = plt.subplot(len(degrees) + 1, trial, (i + 1) * trial + 2)
        polynomial_features = PolynomialFeatures(degree=degrees[i], include_bias=True)
        xx = polynomial_features.fit_transform(x[:, np.newaxis])
        linear_regression1 = LinearRegression()
        linear_regression1.fit(xx, y)
        plt.plot(X_plot, true_fun(X_plot), label="true func")
        plt.plot(X_plot, linear_regression1.predict(polynomial_features.fit_transform(X_plot[:, np.newaxis])), label="model")
        plt.scatter(x, y, s=20, label="samples")
        plt.xlim((0, 1))
        plt.ylim((-1.5, 1.5))
        plt.legend(loc="best", labelcolor='#073642')

        x, y = X[:, 2], Y[:, 2]
        ax = plt.subplot(len(degrees) + 1, trial, (i + 1) * trial + 3)
        polynomial_features = PolynomialFeatures(degree=degrees[i], include_bias=True)
        xx = polynomial_features.fit_transform(x[:, np.newaxis])
        linear_regression2 = LinearRegression()
        linear_regression2.fit(xx, y)
        plt.plot(X_plot, true_fun(X_plot), label="true func")
        plt.plot(X_plot, linear_regression2.predict(polynomial_features.fit_transform(X_plot[:, np.newaxis])), label="model")
        plt.scatter(x, y, s=20, label="samples")
        plt.xlim((0, 1))
        plt.ylim((-1.5, 1.5))
        plt.legend(loc="best", labelcolor='#073642')

        x, y = X[:, 3], Y[:, 3]
        ax = plt.subplot(len(degrees) + 1, trial, (i + 1) * trial + 4)
        polynomial_features = PolynomialFeatures(degree=degrees[i], include_bias=True)
        xx = polynomial_features.fit_transform(x[:, np.newaxis])
        linear_regression3 = LinearRegression()
        linear_regression3.fit(xx, y)
        plt.plot(X_plot, true_fun(X_plot), label="true func")
        plt.plot(X_plot, linear_regression3.predict(polynomial_features.fit_transform(X_plot[:, np.newaxis])), label="model")
        plt.scatter(x, y, s=20, label="samples")
        plt.xlim((0, 1))
        plt.ylim((-1.5, 1.5))
        plt.legend(loc="best", labelcolor='#073642')

        x, y = X[:, 4], Y[:, 4]
        ax = plt.subplot(len(degrees) + 1, trial, (i + 1) * trial + 5)
        polynomial_features = PolynomialFeatures(degree=degrees[i], include_bias=True)
        xx = polynomial_features.fit_transform(x[:, np.newaxis])
        linear_regression3 = LinearRegression()
        linear_regression3.fit(xx, y)
        plt.plot(X_plot, true_fun(X_plot), label="true func")
        plt.plot(X_plot, linear_regression3.predict(polynomial_features.fit_transform(X_plot[:, np.newaxis])), label="model")
        plt.scatter(x, y, s=20, label="samples")
        plt.xlim((0, 1))
        plt.ylim((-1.5, 1.5))
        plt.legend(loc="best", labelcolor='#073642')

    plt.show()
```
