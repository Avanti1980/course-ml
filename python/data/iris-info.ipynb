```python
import seaborn as sns

iris = sns.load_dataset("iris")
iris.head()  # 前5个样本
```

```python
iris.tail()  # 最后5个样本
```

```python
iris.describe()  # 特征统计信息
```

```python
import matplotlib.pyplot as plt

g = sns.pairplot(iris, hue="species", aspect=1, markers=["o", "s", "D"])
plt.show()
```

```python
g = sns.pairplot(iris, hue="species", diag_kind="hist", kind="kde", aspect=1)
plt.show()
```
