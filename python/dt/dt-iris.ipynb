```python
from sklearn.datasets import load_iris

iris = load_iris()
X, y = iris.data, iris.target

X, y
```

```python
kind_dict = {"sepal length (cm)": "花萼长度", "sepal width (cm)": "花萼宽度", "petal length (cm)": "花瓣长度", "petal width (cm)": "花瓣宽度"}
iris.feature_names = list(map(kind_dict.get, iris.feature_names))

kind_dict = {"setosa": "山鸢尾", "versicolor": "杂色鸢尾", "virginica": "维吉尼亚鸢尾"}
iris.target_names = list(map(kind_dict.get, iris.target_names))

iris
```

```python
from sklearn import tree
import graphviz 

clf = tree.DecisionTreeClassifier()
clf = clf.fit(X, y)
dot_data = tree.export_graphviz(clf, out_file=None, feature_names=iris.feature_names, class_names=iris.target_names, filled=True, rounded=True, special_characters=True)
dot_data = dot_data.replace("samples", "样本数")
dot_data = dot_data.replace("class", "类别")
dot_data = dot_data.replace("setosa", "山鸢尾")
dot_data = dot_data.replace("versicolor", "杂色鸢尾")
dot_data = dot_data.replace("virginica", "维吉尼亚鸢尾")
dot_data = dot_data.replace("value", "类别分布")
dot_data = dot_data.replace("sepal length", "花萼长度")
dot_data = dot_data.replace("sepal width", "花萼宽度")
dot_data = dot_data.replace("petal length", "花瓣长度")
dot_data = dot_data.replace("petal width", "花瓣宽度")
dot_data = dot_data.replace("entropy", "熵")
dot_data = dot_data.replace("gini", "基尼指数")
dot_data = dot_data.replace("True", "是")
dot_data = dot_data.replace("False", "否")
graph = graphviz.Source(dot_data) 
graph
```
