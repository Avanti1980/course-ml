```python
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

zhisong = fm.FontEntry(fname="/home/avanti/Fonts/LXGW/LXGWNeoZhiSongScreenFull.ttf", name="LXGW Neo ZhiSong Screen Full")
fm.fontManager.ttflist.insert(0, zhisong)
plt.rcParams.update({
    "font.family": ["LXGW Neo ZhiSong Screen Full"],
    "font.size": 16,
    "axes.grid": True,
})

# 加载数据
iris = sns.load_dataset("iris")
iris.columns = ["花萼长度", "花萼宽度", "花瓣长度", "花瓣宽度", "类别"]
kind_dict = {"setosa": "山鸢尾", "versicolor": "杂色鸢尾", "virginica": "维吉尼亚鸢尾"}
iris["类别"] = iris["类别"].map(kind_dict)

g = sns.pairplot(iris, hue="类别", markers=["o", "s", "D"])

for ax in g.axes.flat:
    ax.tick_params(axis='both', labelsize=10, labelfontfamily="Ysabeau Office")
    
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset("iris")
g = sns.pairplot(iris)

# 正确：所有参数都写成关键字形式
# Seaborn 会自动把 ax 传给 tick_params，然后把 axis='both' 等传进去
g.map(plt.tick_params, axis='both', labelsize=10, color='red')

plt.show()
```
